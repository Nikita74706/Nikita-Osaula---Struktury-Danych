```python
import geopandas as gpd



```

```python
 
pasy = gpd.read_file("D:\\Statystyka przestrzenna\\dane\\pl_pasy_ukształtowania.geojson")

pasy2 = pasy.to_crs("EPSG:2180")

pasy2["area_m2"] = pasy2.geometry.area

print(pasy2["area_m2"])
print(pasy2["area_m2"]/1000000)

```

```python
cities = gpd.read_file("D:\\Statystyka przestrzenna\\dane\\pl_miasta_wojewódzkie.geojson")
cities.set_crs("EPSG:4326", inplace=True)
cities2180 = cities.to_crs("EPSG:2180")

bbox = cities2180.total_bounds
print(bbox) 

from shapely.geometry import box

minx, miny, maxx, maxy = cities2180.total_bounds 
bbox_geom = box(minx, miny, maxx, maxy)
bbox_gdf = gpd.GeoDataFrame(geometry=[bbox_geom],crs=cities2180.crs)

obwod = bbox_gdf.length
print(obwod.iloc[0]/1000) 
```

```python
from shapely import get_num_coordinates
rivers = gpd.read_file("D:\\Statystyka przestrzenna\\dane\\pl_rzeki.geojson")
rivers.set_crs("EPSG:4326", inplace=True)
rivers2180 = rivers.to_crs("EPSG:2180")
#print(rivers2180.crs)
rivers2180["liczba linii"] = rivers2180.count_geometries()
print(rivers2180["liczba linii"])
rivers2180["punkty"] = get_num_coordinates(rivers2180.geometry)
print(rivers2180["punkty"])
```
